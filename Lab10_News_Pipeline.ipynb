```python
def report_agent(topic, summary):

    return llm.invoke(

        f'Create a short Markdown report titled "News: {topic}". '

        f'Use a heading and these points:\n{summary}').content
```

```python
def search_agent(topic):
    print("Agent 1: Searching...")
    return search.invoke(topic)


def summariser_agent(raw):
    print("Agent 2: Summarising...")

    prompt = f"""
    Summarise the following search results.
    Keep only the most important information.

    Search results:
    {raw}
    """

    response = llm.invoke(prompt)
    return response.content


def report_agent(topic, summary):
    print("Agent 3: Creating report...")

    prompt = f"""
    Write a professional news report about: {topic}

    Use the following summary:
    {summary}

    Structure the report with:
    # Title
    ## Introduction
    ## Key Developments
    ## Conclusion
    """

    response = llm.invoke(prompt)
    return response.content
```

```python
topic = 'artificial intelligence'



raw = search_agent(topic)          # Agent 1

summary = summariser_agent(raw)    # Agent 2

report = report_agent(topic, summary)  # Agent 3



with open('news_report.md', 'w') as f:

    f.write(report)



print(report)

print('\nSaved as news_report.md (see the Files panel on the left).')
```
