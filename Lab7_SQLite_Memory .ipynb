```python
from langchain_groq import ChatGroq

llm = ChatGroq(api_key="PASTE_YOUR_GROQ_API_KEY_HERE",

               model="llama-3.1-8b-instant", temperature=0)
```

```python
import sqlite3



conn = sqlite3.connect('memory1.db')

conn.execute('CREATE TABLE IF NOT EXISTS facts (key TEXT, value TEXT)')

conn.commit()

print('Database ready: memory1.db')
```

```python
def remember(key, value):

    conn.execute('INSERT INTO facts VALUES (?, ?)', (key, value))

    conn.commit()



remember('name', 'Nagesh')

remember('favourite_subject', 'Chemistry')

print('Saved!')
```

```python
def recall_all():

    rows = conn.execute('SELECT key, value FROM facts').fetchall()

    return '; '.join(f'{k} = {v}' for k, v in rows)



memory = recall_all()

print('Recalled memory:', memory)



reply = llm.invoke(f'Known facts: {memory}. '

                   'Greet the user by name and mention their favourite subject.')

print(reply.content)
```
